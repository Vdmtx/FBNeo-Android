```python
# Nexus Éter – Extração e Tradução de Legendas (4 mídias, seleção de idioma por número)

# 1. Instalações (execute no Colab)
!pip install -q openai-whisper deep-translator ffmpeg-python

import whisper, os
from deep_translator import GoogleTranslator
from google.colab import files

# 2. Criação de pastas
os.makedirs("uploads", exist_ok=True)
os.makedirs("legendas", exist_ok=True)

# 3. Upload de até 4 arquivos
print("📤 Selecione até 4 arquivos (vídeo/áudio):")
uploaded = files.upload()
filenames = list(uploaded.keys())[:4]

# 4. Lista de idiomas e códigos
langs_list = [
    ("Português", "pt"),
    ("Inglês",    "en"),
    ("Espanhol",  "es"),
    ("Alemão",    "de"),
    ("Francês",   "fr"),
    ("Chinês",    "zh-cn"),
    ("Russo",     "ru"),
    ("Japonês",   "ja"),
    ("Árabe",     "ar"),
    ("Vietnamita","vi"),
    ("Coreano",   "ko"),
    ("Tagalo",    "tl"),
    ("Cebuano",   "ceb"),
]

# 5. Mostra opções numeradas
print("\n🌐 Escolha o idioma para cada arquivo:")
for idx, (name, code) in enumerate(langs_list, start=1):
    print(f"  {idx:2d}. {name} ({code})")

# 6. Pergunta número para cada vídeo
selection = {}
for f in filenames:
    num = input(f"> idioma para '{f}' (1-{len(langs_list)}): ").strip()
    try:
        choice = int(num)
        if 1 <= choice <= len(langs_list):
            selection[f] = langs_list[choice-1][1]
        else:
            raise ValueError
    except:
        print("  ⚠️ inválido, usando Português.")
        selection[f] = "pt"

# 7. Formata tempo para SRT
def fmt_tempo(s):
    h=int(s//3600); m=int((s%3600)//60)
    sec=int(s%60); ms=int((s-int(s))*1000)
    return f"{h:02}:{m:02}:{sec:02},{ms:03}"

# 8. Carrega o modelo Whisper
model = whisper.load_model("base")

# 9. Processa todos os arquivos
for f in filenames:
    path = f"uploads/{f}"
    with open(path, "wb") as fp:
        fp.write(uploaded[f])
    print(f"\n🎧 Transcrevendo '{f}'...")
    result = model.transcribe(path)
    tgt = selection[f]

    srt = ""
    for seg in result["segments"]:
        srt += f"{seg['id']+1}\n"
        srt += f"{fmt_tempo(seg['start'])} --> {fmt_tempo(seg['end'])}\n"
        text = seg["text"].strip()
        tr = GoogleTranslator(source="auto", target=tgt).translate(text)
        srt += f"{tr}\n\n"

    base = os.path.splitext(f)[0]
    out = f"legendas/{base}.srt"
    with open(out, "w", encoding="utf-8") as fp:
        fp.write(srt)

    # 10. Baixa automaticamente
    print(f"📥 Baixando legenda: {base}.srt")
    files.download(out)
```
