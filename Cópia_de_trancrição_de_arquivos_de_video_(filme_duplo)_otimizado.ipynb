```python
# 🚀 Nexus Éter – Transcrição e Legenda Automática (Dois Arquivos)
!pip install -q git+https://github.com/openai/whisper.git deep-translator moviepy

import whisper, os
from deep_translator import GoogleTranslator
from moviepy.editor import VideoFileClip
from google.colab import files
from datetime import timedelta

def formatar_tempo(segundos):
    tempo = str(timedelta(seconds=int(segundos)))
    milis = int((segundos % 1) * 1000)
    return f"{tempo},{milis:03}".rjust(12, "0")

def processar_video(nome_arquivo, idioma_alvo):
    base = os.path.splitext(nome_arquivo)[0]
    audio_path = f"{base}.wav"

    print(f"\n🎬 Processando: {nome_arquivo}")

    # 🎧 Extração do áudio
    try:
        print("🎧 Extraindo áudio...")
        clip = VideoFileClip(nome_arquivo)
        clip.audio.write_audiofile(audio_path, verbose=False, logger=None)
        print("✅ Áudio extraído.")
    except Exception as e:
        print(f"❌ Erro ao extrair áudio de {nome_arquivo}: {e}")
        return

    # 🧠 Transcrição
    try:
        print("🧠 Transcrevendo com Whisper...")
        model = whisper.load_model("base")
        resultado = model.transcribe(audio_path)
        segmentos = resultado["segments"]
        transcricao = resultado["text"]
        print("✅ Transcrição feita.")
    except Exception as e:
        print(f"❌ Erro na transcrição: {e}")
        return

    # 📝 Legenda original (.srt)
    srt_path = f"{base}.srt"
    txt_path = f"{base}.txt"
    try:
        with open(srt_path, "w", encoding="utf-8") as srt:
            for i, seg in enumerate(segmentos):
                srt.write(f"{i+1}\n")
                srt.write(f"{formatar_tempo(seg['start'])} --> {formatar_tempo(seg['end'])}\n")
                srt.write(f"{seg['text'].strip()}\n\n")
        with open(txt_path, "w", encoding="utf-8") as txt:
            txt.write(transcricao)
        print("✅ Legenda .srt e transcrição .txt geradas.")
    except Exception as e:
        print(f"❌ Erro ao gerar legenda: {e}")
        return

    # 🌍 Tradução
    traduzida_path = f"{base}_{idioma_alvo}.srt"
    try:
        print(f"🌍 Traduzindo legenda para {idioma_alvo}...")
        linhas = []
        with open(srt_path, "r", encoding="utf-8") as original:
            for linha in original:
                if linha.strip().isdigit() or "-->" in linha or linha.strip() == "":
                    linhas.append(linha)
                else:
                    traduzida = GoogleTranslator(source='auto', target=idioma_alvo).translate(linha.strip())
                    linhas.append(traduzida + "\n")
        with open(traduzida_path, "w", encoding="utf-8") as traducao:
            traducao.writelines(linhas)
        print("✅ Tradução completa.")
    except Exception as e:
        print(f"❌ Erro na tradução da legenda: {e}")
        return

    # ⬇️ Download
    print("\n⬇️ Pronto para download:")
    print(f"- 📄 {txt_path}")
    print(f"- 🎞️ {srt_path}")
    print(f"- 🌐 {traduzida_path}")
    files.download(txt_path)
    files.download(srt_path)
    files.download(traduzida_path)


# 🔧 Interface
print("📥 Envie seu primeiro vídeo ou áudio")
upload1 = files.upload()
print("📥 Envie seu segundo vídeo ou áudio")
upload2 = files.upload()

arquivo1 = list(upload1.keys())[0]
arquivo2 = list(upload2.keys())[0]

idiomas = {
    1: 'pt', 2: 'en', 3: 'es', 4: 'ja', 5: 'ru', 6: 'zh',
    7: 'ceb', 8: 'tl', 9: 'de', 10: 'fr', 11: 'it', 12: 'ko'
}
print("\n🌍 Escolha o número do idioma da legenda traduzida:")
for i, code in idiomas.items():
    print(f"{i}. {code}")
escolha = int(input("\nIdioma desejado (número): "))
idioma_alvo = idiomas.get(escolha, 'pt')

# 🔁 Processamento dos dois vídeos
processar_video(arquivo1, idioma_alvo)
processar_video(arquivo2, idioma_alvo)

print("\n✅ Nexus Éter finalizou o processo para os dois vídeos.")
```
