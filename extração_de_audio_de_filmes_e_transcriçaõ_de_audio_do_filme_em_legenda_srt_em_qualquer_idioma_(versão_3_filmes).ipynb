```python
# ✅ TRANSCRIÇÃO E LEGENDA DE ÁUDIO E VÍDEO - NEXUS ÉTER

# 🔧 INSTALAÇÃO DE BIBLIOTECAS NECESSÁRIAS
!pip install -q git+https://github.com/openai/whisper.git deep-translator moviepy tqdm

import subprocess
from google.colab import files
from datetime import timedelta
import moviepy.editor as mp
import whisper
from deep_translator import GoogleTranslator
import os
from tqdm import tqdm_notebook as tqdm

# Função de formatação de tempo para legenda

def format_timestamp(seconds):
    hours, remainder = divmod(seconds, 3600)
    minutes, seconds = divmod(remainder, 60)
    milliseconds = int((seconds % 1) * 1000)
    return f'{int(hours):02}:{int(minutes):02}:{int(seconds):02},{milliseconds:03}'

# Função principal de processamento

def process_video(video_path, target_lang):
    print(f"\n📤 Processando arquivo: {video_path}")
    audio_path = os.path.splitext(video_path)[0] + ".wav"

    try:
        print("🔊 [1/5] Extraindo áudio...")
        clip = mp.VideoFileClip(video_path)
        clip.audio.write_audiofile(audio_path, verbose=False, logger=None)
        print("✅ Áudio extraído.")
    except Exception as e:
        print(f"❌ Erro na extração do áudio: {str(e)}")
        return

    try:
        print("🧠 [2/5] Transcrevendo com Whisper...")
        model = whisper.load_model("base")
        result = model.transcribe(audio_path)
        segments = result['segments']
        transcription = result['text']
        print("✅ Transcrição completa.")
    except Exception as e:
        print(f"❌ Erro na transcrição: {str(e)}")
        return

    output_srt = os.path.splitext(video_path)[0] + '.srt'
    try:
        print("📝 [3/5] Gerando legenda .srt...")
        with open(output_srt, "w", encoding="utf-8") as srt:
            for i, seg in enumerate(segments):
                srt.write(f"{i+1}\n")
                srt.write(f"{format_timestamp(seg['start'])} --> {format_timestamp(seg['end'])}\n")
                srt.write(f"{seg['text'].strip()}\n\n")
        print("✅ Legenda gerada.")
    except Exception as e:
        print(f"❌ Erro ao gerar legenda: {str(e)}")
        return

    translated_filename = os.path.splitext(video_path)[0] + f'_{target_lang}.srt'
    try:
        print(f"🌍 [4/5] Traduzindo legenda para {target_lang}...")
        translated_srt = []
        with open(output_srt, "r", encoding="utf-8") as file:
            for line in file:
                if line.strip().isdigit() or '-->' in line or line.strip() == '':
                    translated_srt.append(line)
                else:
                    translated_line = GoogleTranslator(source='auto', target=target_lang).translate(line)
                    translated_srt.append(translated_line + '\n')
        with open(translated_filename, "w", encoding="utf-8") as file:
            file.writelines(translated_srt)
        print("✅ Tradução concluída.")
    except Exception as e:
        print(f"❌ Erro na tradução da legenda: {str(e)}")
        return

    print("⬇️ [5/5] Preparando arquivos para download...")
    try:
        files.download(output_srt)
        files.download(translated_filename)
        print("✅ Arquivos prontos para download.")
    except Exception as e:
        print(f"❌ Erro no download: {str(e)}")

# Uploads
print("📥 Envie o primeiro vídeo ou áudio")
upload1 = files.upload()
print("📥 Envie o segundo vídeo ou áudio")
upload2 = files.upload()
print("📥 Envie o terceiro vídeo ou áudio")
upload3 = files.upload()

arquivo1 = list(upload1.keys())[0]
arquivo2 = list(upload2.keys())[0]
arquivo3 = list(upload3.keys())[0]

# Idioma
idiomas = {
    1: 'pt', 2: 'en', 3: 'ja', 4: 'ru', 5: 'zh', 6: 'tl',
    7: 'ceb', 8: 'de', 9: 'es', 10: 'fr', 11: 'it', 12: 'ko'
}
print("\n🌐 Escolha o número do idioma para tradução:")
for i, code in idiomas.items():
    print(f"{i}. {code}")

escolha = int(input("\nEscolha o idioma (número): "))
target_lang = idiomas.get(escolha, 'pt')

# Processar os vídeos
process_video(arquivo1, target_lang)
process_video(arquivo2, target_lang)
process_video(arquivo3, target_lang)

print("\n✅ Nexus Éter finalizou o processo para todos os vídeos.")
```
